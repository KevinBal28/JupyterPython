```python
from ultralytics import YOLO
import yaml
#from google.colab import drive
```

## Configuration of Variables and functions
### *Configuración de variables y funciones*

```python
pathDataset = '/content/drive/MyDrive/Datasets/datasetLaPazC'
pathYOLO = f'{pathDataset}/datasetLaPazC/yolo26n.pt'
```

## Code
### *Código*

### Creating a `.yaml` file
#### *Creación de un archivo `.yaml`*

```python
# drive.mount('/content/drive')  # Mount Google Drive in Colab / Montar Google Drive en Colab

datasetConfig = {  # Define dataset configuration dict / Definir dict de configuración del dataset
    'path': pathDataset,  # Dataset root path / Ruta raíz del dataset
    'train': 'images/train',  # Training images relative path / Ruta relativa de imágenes de entrenamiento
    'val': 'images/val',  # Validation images relative path / Ruta relativa de imágenes de validación
    'test': 'images/test',  # Test images relative path / Ruta relativa de imágenes de prueba
    'nc': 7,  # Number of classes / Número de clases
    'names': ['arbol', 'poste', 'persona', 'vehiculo', 'senializacion', 'basurero', 'perro']  # Class names / Nombres de las clases
}


with open('data.yaml', 'w') as f:  # Open (or create) data.yaml for writing / Abrir (o crear) data.yaml para escritura
    yaml.dump(datasetConfig, f)  # Write dataset config as YAML / Escribir configuración del dataset como YAML
```

### Train the model
#### *Entrenar el modelo*

```python
model = YOLO(pathYOLO)  # Load YOLO model from given path / Cargar modelo YOLO desde la ruta dada

results = model.train(  # Train the YOLO model / Entrenar el modelo YOLO
    data = 'data.yaml',  # Dataset config file / Archivo de configuración del dataset
    epochs = 90,  # Number of training epochs / Número de épocas de entrenamiento
    patience = 30,  # Early stopping patience (epochs without improvement) / Paciencia para early stopping (épocas sin mejora)
    project = f"{pathDataset}/runs/detect",  # Output project directory / Directorio del proyecto de salida
    imgsz = 640,  # Input image size / Tamaño de imagen de entrada
    batch = 32,  # Batch size / Tamaño del lote
    device = 0,  # GPU device index / Índice del dispositivo GPU
    augment = True,  # Enable data augmentation / Habilitar aumento de datos
    cos_lr = True,  # Use cosine learning rate scheduler / Usar planificador de LR coseno
    cache = "ram",  # Cache images in RAM / Almacenar imágenes en caché en RAM
    workers = 4,  # Number of dataloader workers / Número de workers del dataloader
    lr0 = 0.01,  # Initial learning rate / Tasa de aprendizaje inicial
    optimizer = 'AdamW',  # Optimizer type / Tipo de optimizador
    label_smoothing = 0.1,  # Label smoothing factor / Factor de suavizado de etiquetas
    # mixup = 0.1,  # MixUp augmentation probability / Probabilidad de aumento MixUp
    # copy_paste = 0.1,  # Copy-paste augmentation probability / Probabilidad de aumento copy-paste
    # amp = True,  # Enable Automatic Mixed Precision / Habilitar precisión mixta automática
    # warmup_epochs = 3,  # Warmup epochs / Épocas de calentamiento
    # warmup_momentum = 0.8,  # Warmup momentum / Momento de calentamiento
    # close_mosaic = 10,  # Disable mosaic last N epochs / Deshabilitar mosaico en las últimas N épocas
    # mosaic = 1.0,  # Mosaic augmentation probability / Probabilidad de aumento mosaic
    # exist_ok = True,  # Allow overwriting existing project dir / Permitir sobrescribir el directorio del proyecto existente
    # verbose = True  # Verbose training output / Salida detallada del entrenamiento
)
```

### Evaluate the model
#### *Evaluar el modelo*

```python
model = YOLO(f"{pathDataset}/runs/detect/train/weights/best.pt")  # Load best trained weights / Cargar mejores pesos entrenados
metrics = model.val(  # Validate the model / Validar el modelo
    data='data.yaml',  # Dataset config file / Archivo de configuración del dataset
    split='test',  # Use test split / Usar split de prueba
    project = f'{pathDataset}/runs/detect',  # Output project directory / Directorio del proyecto de salida
    imgsz = 640, # Input image size / Tamaño de imagen de entrada
    batch = 32,  # Batch size / Tamaño del lote
    device = 0,  # GPU device index / Índice del dispositivo GPU
    )
```
